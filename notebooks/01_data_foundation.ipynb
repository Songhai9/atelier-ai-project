The initial loading check reads the first 1,000 records from each file.

```python
PROJECT_ROOT = Path(atelier_ai.__file__).resolve().parents[2]
RAW_DATA_DIRECTORY = PROJECT_ROOT / "data" / "raw" / "hm"

PREVIEW_ROWS = 1_000

SOURCE_FILENAMES = (
    "articles.csv",
    "customers.csv",
    "transactions_train.csv",
)

for filename in SOURCE_FILENAMES:
    path = RAW_DATA_DIRECTORY / filename
    if not path.is_file():
        raise FileNotFoundError(f"Required source file not found: {path}")

print("Raw data directory found.")
```

### 3.2. Limited Reads

All fields are initially read as strings to preserve their lexical

Automatic missing-value recognition is disabled for this inspection.
Empty fields therefore remain empty strings. Numerical, categorical and

```python
previews = {
    Path(filename).stem: read_source_csv(
        RAW_DATA_DIRECTORY / filename,
        nrows=PREVIEW_ROWS,
    )
    for filename in SOURCE_FILENAMES
}
```

### 3.3. Loading Summary

The summary reports loaded dimensions and estimated DataFrame memory
usage for each preview. Loaded row counts are not full-file row counts.

Published outputs contain structural diagnostics rather than source

```python
summary_rows = []

for dataset_name, frame in previews.items():
    summary_rows.append(
        {
            "dataset": dataset_name,
            "rows_loaded": len(frame),
            "columns_loaded": frame.shape[1],
            "memory_mib": (
                frame.memory_usage(index=True, deep=True).sum() / 1024**2
            ),
        }
    )

loading_summary = pd.DataFrame(summary_rows)
loading_summary.round({"memory_mib": 3})
```

### 3.4. Full-File Reading Strategy

All three files are scanned in chunks of up to 100,000 records.
Chunks are processed sequentially rather than accumulated.

Article and customer fields remain strings pending detailed inspection.

For transactions:
- customer, article and sales-channel identifiers remain strings;
- dates are parsed using the YYYY-MM-DD format;
- prices are converted to nullable Float64 values;
- blank dates and prices are treated as missing;
- invalid non-blank values raise conversion errors.

Source files are not modified. No price rescaling or currency
interpretation is applied.

```python
from time import perf_counter

CHUNK_ROWS = 100_000
```

### 3.5. Full-File Scan

The scan reports parsed record counts, chunk counts, elapsed time and
the largest observed processed-chunk memory estimate.

Chunk memory is measured after the selected conversions. It is not a
measurement of total process memory or peak memory during parsing.

```python
scan_rows = []
transaction_type_summary = None

for filename in SOURCE_FILENAMES:
    dataset_name = Path(filename).stem
    rows_read = 0
    chunks_read = 0
    max_chunk_mib = 0.0
    started_at = perf_counter()

    print(f"Scanning {filename}...", flush=True)

    with read_source_chunks(
        RAW_DATA_DIRECTORY / filename,
        chunk_rows=CHUNK_ROWS,
    ) as reader:
        for chunk in reader:
            if dataset_name == "transactions_train":
                chunk["t_dat"] = pd.to_datetime(
                    chunk["t_dat"].replace("", pd.NA),
                    format="%Y-%m-%d",
                    errors="raise",
                )

                chunk["price"] = pd.to_numeric(
                    chunk["price"].replace("", pd.NA),
                    errors="raise",
                ).astype("Float64")

            rows_read += len(chunk)
            chunks_read += 1

            chunk_mib = (
                chunk.memory_usage(index=True, deep=True).sum()
                / 1024**2
            )
            max_chunk_mib = max(max_chunk_mib, chunk_mib)

            if dataset_name == "transactions_train" and chunks_read == 1:
                transaction_type_summary = pd.DataFrame(
                    {
                        "column": chunk.columns,
                        "dtype": [
                            str(dtype) for dtype in chunk.dtypes
                        ],
                    }
                )

            if chunks_read % 50 == 0:
                print(
                    f"  {rows_read:,} records processed",
                    flush=True,
                )

            del chunk

    scan_rows.append(
        {
            "dataset": dataset_name,
            "rows_read": rows_read,
            "chunks_read": chunks_read,
            "max_chunk_mib": max_chunk_mib,
            "elapsed_seconds": perf_counter() - started_at,
        }
    )

    print(f"Completed: {rows_read:,} records.", flush=True)

scan_summary = pd.DataFrame(scan_rows)
```

```python
scan_summary.round(
    {
        "max_chunk_mib": 2,
        "elapsed_seconds": 2,
    }
)
```

### 6.3. Chronological Evaluation Windows

The calendar is loaded from `configs/temporal.yaml`.

Intervals use an inclusive start and an exclusive end: [start, end).
Pre-validation history is a source pool, not the supervised training table.

Six historical training target weeks precede validation.
Validation, intermediate V1 evaluation and final confirmation each
cover seven days.

Development exposure and interpretation limits are documented in the
[evaluation strategy](../docs/evaluation.md).
Final-week recommendation performance remains reserved.

Transaction volumes below are source-coverage diagnostics.
They are not recommendation-performance metrics.

```python
protocol = load_temporal_protocol(
    PROJECT_ROOT / "configs" / "temporal.yaml"
)

source_start = daily_activity["date"].min()
source_end_exclusive = (
    daily_activity["date"].max() + pd.Timedelta(days=1)
)

if source_end_exclusive != protocol["confirmation_end_exclusive"]:
    raise ValueError("Source coverage differs from the configured end date.")

if not source_start < protocol["validation_start"]:
    raise ValueError("No pre-validation history is available.")

period_specs = [
    ("pre_validation_history",
     source_start, protocol["validation_start"]),
    ("validation_target",
     protocol["validation_start"], protocol["v1_evaluation_start"]),
    ("v1_evaluation_target",
     protocol["v1_evaluation_start"], protocol["confirmation_start"]),
    ("confirmation_target",
     protocol["confirmation_start"], protocol["confirmation_end_exclusive"]),
]

window_rows = []

for name, start, end in period_specs:
    period = daily_activity.loc[
        interval_mask(daily_activity["date"], start, end)
    ]

    window_rows.append({
        "period": name,
        "start_inclusive": start.date(),
        "end_exclusive": end.date(),
        "calendar_days": len(period),
        "days_with_records": int(period["observed_in_source"].sum()),
        "transaction_rows": int(period["transaction_rows"].sum()),
    })

window_summary = pd.DataFrame(window_rows)

if window_summary["calendar_days"].sum() != len(daily_activity):
    raise ValueError("Evaluation windows do not partition the calendar.")

if (
    window_summary["transaction_rows"].sum()
    != daily_activity["transaction_rows"].sum()
):
    raise ValueError("Evaluation-window transaction counts do not reconcile.")

if not window_summary["calendar_days"].iloc[1:].eq(
    protocol["horizon_days"]
).all():
    raise ValueError("A target window differs from the configured horizon.")

training_schedule = make_training_schedule(
    validation_cutoff=protocol["validation_start"],
    n_weeks=protocol["training_weeks"],
    history_days=protocol["history_days"],
    horizon_days=protocol["horizon_days"],
)

print("Protocol:", protocol["protocol_id"])
window_summary
```

### 6.5. Temporal Leakage Controls

For every prediction cutoff t:

- candidates and transaction-derived features use observations before t;
- relevance contains distinct purchased articles in [t, t + 7 days);
- training targets used for selection end no later than validation starts;
- learned parameters and preprocessing use training examples during selection;
- methods and hyperparameters are selected using validation MAP@12;
- a selected learned model is refitted on training plus validation only
  after modelling choices and the iteration budget are frozen;
- V1 evaluation and confirmation targets are excluded from that refit;
- histories and candidates are refreshed before each prediction cutoff;
- purchases inside a target week do not update its initial predictions.

The final LightGBM/PyTorch comparison and fitting procedures must be
frozen before confirmation performance is examined.

Development exposure and its interpretation limits are documented in
the evaluation strategy.

Metadata features require temporal evidence or an explicit stability
assumption. All target purchases remain in evaluation, including those
absent from candidates.

These historical weeks do not establish robustness across seasons
or measurable business impact.
